```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set the style
sns.set_theme(style="whitegrid")

# Create Plot 2
plt.figure(figsize=(12, 6))
sns.countplot(
    data=Data, 
    x='region', 
    order=Data['region'].value_counts().index, 
    palette='magma',hue = 'region',
)

plt.title('Distribution by Region', fontsize=15)
plt.xlabel('Region')
plt.ylabel('Count')
plt.xticks(rotation=45) # Rotates the text so names don't overlap
# plt.savefig('bargraph1.png')
# plt.close()
plt.show()
```

```python
# Define the bins and labels
bins = [0, 25, 50, 75, 100]
labels = ['0-25%', '26-50%', '51-75%', '76-100%']

# Create a new column for the remote work categories
Data['remote_work_category'] = pd.cut(Data['remote_work_percentage'], bins=bins, labels=labels, include_lowest=True)

# Group by the new category and calculate the mean productivity
remote_productivity = Data.groupby('remote_work_category')['productivity_change_percent'].mean().reset_index()

# Display the result
print("Average Productivity Change by Remote Work Level:")
print(remote_productivity)
```

```python
plt.figure(figsize=(10, 6))
sns.barplot(
    data=remote_productivity, 
    x='remote_work_category', 
    y='productivity_change_percent', 
    palette='coolwarm',hue='remote_work_category'
)

plt.title('Impact of Remote Work on Productivity', fontsize=15)
plt.xlabel('Remote Work Percentage')
plt.ylabel('Average Productivity Change (%)')
plt.show()
```

```python

# 1. Calculate the mean productivity for each age group
age_impact = Data.groupby('company_age_group')['productivity_change_percent'].mean().sort_values(ascending=False).reset_index()

# 2. Create the Bar Chart
plt.figure(figsize=(10, 6))
sns.barplot(
    data=age_impact, 
    x='company_age_group', 
    y='productivity_change_percent', 
    palette='magma',hue='company_age_group'
)

plt.title('Average Productivity Change by Company Age Group', fontsize=15)
plt.xlabel('Company Age Group')
plt.ylabel('Avg Productivity Change (%)')
plt.show()
```

```python

# 1. Calculate the percentage of 'Yes' for each industry
# We filter for 'Yes', count them, and divide by the total companies in that industry
ethics_data = Data[Data['ai_ethics_committee'] == 'Yes'].groupby('industry').size() / Data.groupby('industry').size() * 100
ethics_data = ethics_data.sort_values(ascending=False).reset_index(name='percentage')

# 2. Create the Bar Chart
plt.figure(figsize=(10, 6))
sns.barplot(data=ethics_data, x='percentage', y='industry',hue='industry', palette='Reds_r')

plt.title('Percentage of Companies with AI Ethics Committees', fontsize=15)
plt.xlabel('Percentage (%)')
plt.ylabel('Industry')
plt.show()
```

```python
# 1. Calculate average productivity change per industry
prod_by_industry = Data.groupby('industry')['productivity_change_percent'].mean().sort_values(ascending=False).reset_index()

# 2. Create the Bar Chart
plt.figure(figsize=(10, 6))
sns.barplot(data=prod_by_industry, x='productivity_change_percent', y='industry', palette='Blues_r',hue='industry')

plt.title('Average Productivity Change by Industry', fontsize=15)
plt.xlabel('Avg Productivity Change (%)')
plt.ylabel('Industry')
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Prepare the data cleanly
# We aggregate the mean and reset the index
time_series_data = Data.groupby(['survey_year', 'quarter'])['productivity_change_percent'].mean().reset_index()

# 2. Create a clean Label column (e.g., "2023 Q1")
time_series_data['display_period'] = time_series_data['survey_year'].astype(str) + " " + time_series_data['quarter']

# 3. Plotting
plt.figure(figsize=(12, 6))

# Note: We use the 'display_period' for the X-axis
sns.lineplot(
    data=time_series_data, 
    x='display_period', 
    y='productivity_change_percent', 
    marker='o', 
    linewidth=2.5, 
    color='teal'
)

plt.title('Productivity Trend (2023 - 2026)', fontsize=15)
plt.xlabel('Year and Quarter', fontsize=12)
plt.ylabel('Average Productivity Change (%)', fontsize=12)

# This part is key: it keeps the labels short and readable
plt.xticks(rotation=45) 

plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```
